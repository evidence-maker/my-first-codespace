```python
import pypdf

# Path to your PDF file
pdf_path = '/content/Data Analytics Tasks & Instructions — CodeAlpha M1.pdf'

try:
    # Open the PDF file in read-binary mode
    with open(pdf_path, 'rb') as file:
        reader = pypdf.PdfReader(file)

        # Get the number of pages
        num_pages = len(reader.pages)
        print(f'The PDF has {num_pages} pages.')

        # Read the text from the first page as an example
        if num_pages > 0:
            first_page = reader.pages[0]
            text = first_page.extract_text()
            print('\n--- Content of the first page ---\n')
            print(text[:500]) # Print first 500 characters to avoid overwhelming output
            print('\n--- End of first page content (truncated) ---\n')
        else:
            print('The PDF file is empty or has no readable pages.')

except FileNotFoundError:
    print(f'Error: The file {pdf_path} was not found. Please ensure the path is correct.')
except Exception as e:
    print(f'An error occurred: {e}')
```
